```python
#Q1
# given example
import numpy as np

v1 = np.array([1, 2])
v2 = np.array([3, 4])

matrix = np.column_stack((v1, v2))
det = np.linalg.det(matrix)

print("Determinant =", det)
if det != 0:
    print("Vectors are linearly independent")
else:
    print("Vectors are linearly dependent")


# user defined func.
def check_independence(v1, v2):
    matrix = np.column_stack((v1, v2))
    det = np.linalg.det(matrix)
    print("Determinant =", det)
    if det != 0:
        print("Vectors are linearly independent")
    else:
        print("Vectors are linearly dependent")

v1 = np.array([5, 0])
v2 = np.array([0, 5])

check_independence(v1, v2)


# user input
import numpy as np

def lin_dep(m):
    m = np.array(m)
    determinant = np.linalg.det(m)
    if np.isclose(determinant, 0):
        print("The set of vectors are linearly dependent.")
    else:
        print(f"Determinant = {determinant}")
        print("The set of vectors are linearly independent.")

n = int(input("Enter number of vectors: "))
r = int(input("Enter dimension of each vector: "))

m = []
for i in range(n):
    v = []
    print(f"Enter elements of vector {i+1}:")
    for j in range(r):
        v.append(float(input()))
    m.append(v)

lin_dep(m)
```

```python
# Q2
# given example
import numpy as np

v1 = np.array([1, 2])
v2 = np.array([3, 4])
v3 = np.array([1, 0])

matrix = np.column_stack((v1, v2, v3))
rank = np.linalg.matrix_rank(matrix)

print("Rank =", rank)
if rank == 2:
    print("Vectors are linearly independent")
else:
    print("Vectors are linearly dependent")


# user defined func.
def check_rank(v1, v2, v3):
    matrix = np.column_stack((v1, v2, v3))
    rank = np.linalg.matrix_rank(matrix)
    print("Rank =", rank)
    if rank == 2:
        print("Vectors are linearly independent")
    else:
        print("Vectors are linearly dependent")

v1 = np.array([2, 1])
v2 = np.array([4, 2])
v3 = np.array([6, 3])

check_rank(v1, v2, v3)


# user input
import numpy as np

def check_rank(m):
    m = np.array(m)
    rank = np.linalg.matrix_rank(m)
    print("Rank =", rank)
    if rank == m.shape[0]:
        print("Vectors are linearly independent")
    else:
        print("Vectors are linearly dependent")

n = int(input("Enter number of vectors: "))
r = int(input("Enter dimension of each vector: "))

m = []
for i in range(n):
    v = []
    print(f"Enter elements of vector {i+1}:")
    for j in range(r):
        v.append(float(input()))
    m.append(v)

m = np.array(m).T
check_rank(m)
```

```python
#Q3
# given example
import numpy as np

v1 = np.array([1,2,3])
v2 = np.array([2,4,6])
v3 = np.array([3,6,9])

matrix = np.column_stack((v1, v2, v3))
rank = np.linalg.matrix_rank(matrix)

print("Rank =", rank)
if rank < 3:
    print("Vectors are linearly dependent")
else:
    print("Vectors are linearly independent")


# user defined func.
def check_dep(vectors):
    matrix = np.column_stack(vectors)
    rank = np.linalg.matrix_rank(matrix)
    print("Rank =", rank)
    if rank < len(vectors):
        print("Vectors are linearly dependent")
    else:
        print("Vectors are linearly independent")

v1 = np.array([1,0,1])
v2 = np.array([2,1,3])
v3 = np.array([3,1,4])

check_dep([v1, v2, v3])


# user input
import numpy as np

def check_dep(m):
    m = np.array(m)
    m = m.T
    rank = np.linalg.matrix_rank(m)
    print("Rank =", rank)
    if rank < m.shape[1]:
        print("Vectors are linearly dependent")
    else:
        print("Vectors are linearly independent")

n = int(input("Enter number of vectors: "))
r = int(input("Enter dimension of each vector: "))

m = []
for i in range(n):
    v = []
    print(f"Enter elements of vector {i+1}:")
    for j in range(r):
        v.append(float(input()))
    m.append(v)

check_dep(m)
```

```python
#Q4
# given example
import numpy as np

v1 = np.array([1,0,0])
v2 = np.array([0,1,0])
v3 = np.array([0,0,1])

matrix = np.column_stack((v1, v2, v3))
rank = np.linalg.matrix_rank(matrix)

print("Rank =", rank)
if rank == len([v1, v2, v3]):
    print("Vectors are linearly independent")
else:
    print("Vectors are linearly dependent")


# user defined func.
def check_ind(vectors):
    matrix = np.column_stack(vectors)
    rank = np.linalg.matrix_rank(matrix)
    print("Rank =", rank)
    if rank == len(vectors):
        print("Vectors are linearly independent")
    else:
        print("Vectors are linearly dependent")

v1 = np.array([9,0,9])
v2 = np.array([6,2,0])
v3 = np.array([0,6,8])

check_ind([v1, v2, v3])


# user input
import numpy as np

def check_ind(m):
    m = np.array(m)
    m = m.T
    rank = np.linalg.matrix_rank(m)
    print("Rank =", rank)
    if rank == m.shape[1]:
        print("Vectors are linearly independent")
    else:
        print("Vectors are linearly dependent")

n = int(input("Enter number of vectors: "))
r = int(input("Enter dimension of each vector: "))

m = []
for i in range(n):
    v = []
    print(f"Enter elements of vector {i+1}:")
    for j in range(r):
        v.append(float(input()))
    m.append(v)

check_ind(m)
```

```python
#Q5
# Rank = number of linearly independent vectors
# If rank == number of vectors → linearly independent
# If rank < number of vectors → linearly dependent

import numpy as np
from numpy.linalg import matrix_rank

def check(vectors):
    m = np.array(vectors)
    rank = matrix_rank(m)

    print("Vectors:")
    print(m)
    print("Rank:", rank)
    print("Number of vectors:", len(vectors))

    if rank == len(vectors):
        print("Result: Vectors are Linearly Independent\n")
    else:
        print("Result: Vectors are Linearly Dependent\n")


# Example 1: Linearly Independent
vectors1 = [
    [1, 0, 0],
    [0, 1, 0],
    [0, 0, 1]
]

check(vectors1)

# Example 2: Linearly Dependent
vectors2 = [
    [1, 2, 3],
    [2, 4, 6],
    [3, 6, 9]
]

check(vectors2)
```

```python
#TASK 1
import pandas as pd
import numpy as np
from numpy.linalg import matrix_rank

df = pd.read_csv("dataset_1.csv")

print("Dataset 1:")
print(df)

X = df.values
rank_X = matrix_rank(X)

print("\nNumber of features:", df.shape[1])
print("Rank of dataset:", rank_X)

redundant_features = []

for col in df.columns:
    temp_df = df.drop(columns=[col])
    temp_rank = matrix_rank(temp_df.values)

    if temp_rank == rank_X:
        redundant_features.append(col)

print("\nRedundant Feature(s):", redundant_features)
```
