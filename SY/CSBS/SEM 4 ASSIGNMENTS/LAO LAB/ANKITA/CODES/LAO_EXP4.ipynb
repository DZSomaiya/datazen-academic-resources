```python
#Q2
import pandas as pd
import numpy as np

data = pd.read_csv('exp4_gram_schmidt_dataset.csv')
X = data.values
cols = X.shape[1]

for i in range(cols):
    for j in range(i + 1, cols):
        dot = np.dot(X[:, i], X[:, j])
        if dot != 0:
            print("Dataset features are not orthogonal")
            print(f"Dot product of Feature {i+1} and Feature {j+1} = {dot}")
            exit()

print("Dataset features are orthogonal")
```

```python
#Q3
import pandas as pd

data = pd.read_csv('exp4_gram_schmidt_dataset.csv')
corr = data.corr()

print("Correlation matrix:\n")
print(corr)

print("\nHighly correlated feature pairs:\n")

cols = corr.columns

for i in range(len(cols)):
    for j in range(i + 1, len(cols)):
        if abs(corr.iloc[i, j]) > 0.7:
            print(f"{cols[i]} and {cols[j]} are highly correlated")
```

```python
#Q5
import numpy as np

def gram_schmidt(vectors):
    ortho = []

    for v in vectors:
        u = v.copy()

        for o in ortho:
            if np.dot(o, o) != 0:
                u = u - (np.dot(v, o) / np.dot(o, o)) * o

        if np.linalg.norm(u) != 0:
            ortho.append(u)

    return ortho


n = int(input("Enter number of vectors: "))
m = int(input("Enter dimension of vectors: "))

vectors = []

for i in range(n):
    print(f"Enter elements of vector {i+1}:")
    temp = list(map(float, input().split()))

    if len(temp) != m:
        print("Incorrect number of elements entered.")
        exit()

    vectors.append(np.array(temp))

orthogonal_vectors = gram_schmidt(vectors)

print("\nOrthogonal vectors:")
for v in orthogonal_vectors:
    print(v)
```
