```python
import numpy as np
v1 = np.array([2,4])
v2 = np.array([6,8])
v3 = np.array([10,12])

print("--Vector Space Axioms--")
print("Closure property under addition:", v1 + v2)
print("Commutative property under addition v1+v2=v2+v1:", np.array_equal(v1+v2,v2+v1))
print ("Associativity property under addition (v1+v2)+v3 = v1+(v2+v3):", np.array_equal((v1+v2)+v3,v1+(v2+v3)))
print("Identity property under addition v1 + 0 = v1", np.array_equal(v1+0,v1))
```

```python

import numpy as np
x = np.array([4,8,0])
y = np.array([22,11,0])
z = 9
zero = np.array([0,0,0])

print("--Conditions for Subspace--")
print("Closure under addition:", x+y)
print("Closure under scalar multiplication:",z*x)
print("Identity x+0 = x",np.array_equal(x+0,x))
print("Identity:", zero)
print("Thus proved, the set of all vectors of the form (x,y,0) in R^3 is a a subspace.")
```

```python

import numpy as np

def set(v):
    return v[0] * v[1] == 0

v1 = np.array([20, 0])
v2 = np.array([0, 50])
a = 10
zero = np.array([0, 0])

if set(zero):
    print("Zero vector is in the set.")
else:
    print("Zero vector is not in the set.")

mul = a * v1
if set(mul):
    print("Closed under scalar multiplication:", mul)
else:
    print("Not closed under scalar multiplication:", mul)

sum = v1 + v2
if set(sum):
    print("Closed under addition:", sum)
    print("The set is a subspace!")
else:
    print("Not closed under addition:", sum)
    print("The set is not a subspace!")
```

```python

import numpy as np

A = np.array([[1, 2], [2, 4]])

x1 = np.array([3, -1])
x2 = np.array([4, -2])

zero = np.zeros(2)

if np.allclose(A @ x1, zero) and np.allclose(A @ x2, zero):
    print("Both vectors are solutions")
    print("A(x1 + x2) =", A @ (x1 + x2))
    print("A(3x1) =", A @ (3 * x1))
    print("Hence solution space forms a subspace")
else:
    print("Vectors are not valid solutions")
```
